```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/ott_imdb_sql_ready.csv"
)

genre_data = df[
    [
        "title",
        "genres"
    ]
].dropna(
    subset=["genres"]
).copy()

genre_data.head()
```

```python
# Split the genres into a list of individual genres

genre_data["genres"] = (
    genre_data["genres"]
    .astype(str)
    .str.split(",")
)
```

```python
# Explode the genres into separate rows

genre_data = genre_data.explode(
    "genres"
)
```

```python
# Remove leading and trailing whitespace from the genres

genre_data["genres"] = (
    genre_data["genres"]
    .str.strip()
)

genre_data.head(20)
```

```python
print(genre_data.shape)
print(genre_data.head(20))
print(genre_data["genres"].unique()[:20])
```

```python
# Remove Duplicate Relationships

genre_data = genre_data.drop_duplicates()
```

```python
print(
    "Genre relationships:",
    len(genre_data)
)
```

```python
# Create Unique Genre List

genres = (
    genre_data[
        ["genres"]
    ]
    .drop_duplicates()
    .sort_values("genres")
    .reset_index(drop=True)
)

genres.head(20)
```

```python
# Save Genre CSV

genres.to_csv(
    "../data/processed/genres.csv",
    index=False
)
```

```python
# Save Genre Relationships

genre_data.to_csv(
    "../data/processed/title_genre_relationships.csv",
    index=False
)
```

```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/ott_imdb_sql_ready.csv"
)

staging_df = df[
    [
        "platform",
        "title",
        "genres"
    ]
].dropna(
    subset=["platform", "title", "genres"]
).copy()
```

```python
staging_df["genres"] = (
    staging_df["genres"]
    .astype(str)
    .str.split(",")
)
```

```python
staging_df = staging_df.explode(
    "genres"
)
```

```python
staging_df["genres"] = (
    staging_df["genres"]
    .str.strip()
)
```

```python
staging_df.rename(
    columns={
        "genres": "genre_name"
    },
    inplace=True
)
```

```python
staging_df = staging_df.drop_duplicates(
    subset=[
        "platform",
        "title",
        "genre_name"
    ]
).reset_index(drop=True)
```

```python
print("Genre relationships:", len(staging_df))
print(staging_df.head(20))
```

```python
staging_df.to_csv(
    "../data/processed/genre_staging.csv",
    index=False
)

print("genre_staging.csv created successfully!")
```

```python
sql_values = ",\n".join(
    "("
    f"'{str(row.platform).replace(chr(39), chr(39)+chr(39))}', "
    f"'{str(row.title).replace(chr(39), chr(39)+chr(39))}', "
    f"'{str(row.genre_name).replace(chr(39), chr(39)+chr(39))}'"
    ")"
    for row in staging_df.itertuples(index=False)
)

sql_script = (
    "INSERT INTO genre_staging "
    "(platform, title, genre_name)\nVALUES\n"
    + sql_values
    + ";"
)

print(sql_script[:5000])
```

```python
with open(
    "../data/processed/genre_staging_insert.sql",
    "w",
    encoding="utf-8"
) as f:
    f.write(sql_script)

print("SQL file created!")
```
