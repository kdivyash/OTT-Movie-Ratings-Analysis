## 1. Load IMDb Title Basics Dataset

```python
basics_path = "../data/raw/title.basics.tsv.gz"

basics = pd.read_csv(
    basics_path,
    sep="\t",
    na_values="\\N",
    low_memory=False
)

print("Dataset loaded successfully!")
print("Shape:", basics.shape)
```

```python
basics.head()
```

```python
basics.columns.tolist()
```

```python
basics.info()

basics.shape

basics.isnull().sum()
```

```python
basics["titleType"].value_counts()
```

```python
content_types = [
    "movie",
    "tvSeries",
    "tvMiniSeries"
]

basics_filtered = basics[
    basics["titleType"].isin(content_types)
].copy()

print("Filtered Shape:", basics_filtered.shape)

basics_filtered["titleType"].value_counts()
```

```python
basic_columns = [
    "tconst",
    "titleType",
    "primaryTitle",
    "originalTitle",
    "isAdult",
    "startYear",
    "endYear",
    "runtimeMinutes",
    "genres"
]

basics_filtered = basics_filtered[basic_columns].copy()

basics_filtered.head()
```

```python
basics_filtered.rename(
    columns={
        "tconst": "Title_ID",
        "titleType": "Content_Type",
        "primaryTitle": "Title",
        "originalTitle": "Original_Title",
        "isAdult": "Is_Adult",
        "startYear": "Release_Year",
        "endYear": "End_Year",
        "runtimeMinutes": "Runtime_Minutes",
        "genres": "Genres"
    },
    inplace=True
)

basics_filtered.columns
```

```python
numeric_columns = [
    "Release_Year",
    "End_Year",
    "Runtime_Minutes"
]

for column in numeric_columns:
    basics_filtered[column] = pd.to_numeric(
        basics_filtered[column],
        errors="coerce"
    )

basics_filtered.dtypes
```

Let's initially use 1980 onward.

```python
basics_filtered = basics_filtered[
    basics_filtered["Release_Year"] >= 1980
].copy()

basics_filtered["Release_Year"].min()
```

Remove Adult Titles

```python
basics_filtered["Is_Adult"].value_counts()

basics_filtered = basics_filtered[
    basics_filtered["Is_Adult"] == 0
].copy()

basics_filtered["Is_Adult"].value_counts()

```

Clean Runtime (Runtime can contain missing values.)

```python
basics_filtered["Runtime_Minutes"].isnull().sum()
```

Clean Genres

```python
basics_filtered["Genres"].head(10)

```

```python
basics_filtered["Genres"].isnull().sum()
```

Check Ratings Dataset

```python
ratings

ratings.head()

ratings.info()
```

```python
# Convert the numeric fields:

ratings["averageRating"] = pd.to_numeric(
    ratings["averageRating"],
    errors="coerce"
)

ratings["numVotes"] = pd.to_numeric(
    ratings["numVotes"],
    errors="coerce"
)

# Rename Rating Columns

ratings.rename(
    columns={
        "tconst": "Title_ID",
        "averageRating": "IMDb_Rating",
        "numVotes": "IMDb_Votes"
    },
    inplace=True
)

ratings.head()
```

```python
# Check Duplicate IDs
# Before merging, check whether Title_ID is unique.

print(
    "Duplicate Title IDs:",
    ratings["Title_ID"].duplicated().sum()
)

# Do the same for our basics data:

print(
    "Duplicate Title IDs:",
    basics_filtered["Title_ID"].duplicated().sum()
)
```

# Merge the Two Datasets 🔗

```python
df = basics_filtered.merge(
    ratings,
    on="Title_ID",
    how="inner"
)

print("Final merged shape:", df.shape)

df.head()
```

Apply Rating Quality Filter

```python
df_analysis = df[
    df["IMDb_Votes"] >= 100
].copy()

print("Before filter:", df.shape)
print("After filter:", df_analysis.shape)
```

Check Rating Range

```python
df_analysis["IMDb_Rating"].describe()

print(
    "Minimum Rating:",
    df_analysis["IMDb_Rating"].min()
)

print(
    "Maximum Rating:",
    df_analysis["IMDb_Rating"].max()
)

print(
    "Average Rating:",
    df_analysis["IMDb_Rating"].mean()
)
```

Check Release Years

```python
df_analysis["Release_Year"].describe()

df_analysis["Release_Year"].min()
df_analysis["Release_Year"].max()
```

Check Content Distribution

```python
df_analysis["Content_Type"].value_counts()

content_distribution = (
    df_analysis["Content_Type"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

content_distribution
```

```python
# Creating Rating Category

def rating_category(rating):
    if rating >= 9:
        return "Masterpiece"
    elif rating >= 8:
        return "Excellent"
    elif rating >= 7:
        return "Good"
    elif rating >= 6:
        return "Average"
    else:
        return "Low Rated"

df_analysis["Rating_Category"] = (
    df_analysis["IMDb_Rating"]
    .apply(rating_category)
)

df_analysis["Rating_Category"].value_counts()
```

```python
# Create Decade Column

df_analysis["Decade"] = (
    (df_analysis["Release_Year"] // 10) * 10
)

df_analysis[
    ["Title", "Release_Year", "Decade"]
].head(10)
```

```python
# Create Popularity Category

def popularity_category(votes):
    if votes >= 100000:
        return "Very High"
    elif votes >= 10000:
        return "High"
    elif votes >= 1000:
        return "Medium"
    else:
        return "Low"

df_analysis["Popularity_Category"] = (
    df_analysis["IMDb_Votes"]
    .apply(popularity_category)
)

df_analysis["Popularity_Category"].value_counts()
```

```python
df_analysis.info()

df_analysis.isnull().sum()

df_analysis.head()
```

```python
# Sort the DataFrame by IMDb_Rating and IMDb_Votes in descending order, and select the top 20 entries.

top_rated = (
    df_analysis
    .sort_values(
        by=["IMDb_Rating", "IMDb_Votes"],
        ascending=[False, False]
    )
    .head(20)
)

top_rated[
    [
        "Title",
        "Content_Type",
        "Release_Year",
        "Genres",
        "IMDb_Rating",
        "IMDb_Votes"
    ]
]
```

```python
# Save the processed dataset to a CSV file in the "data/processed" directory.

import os

os.makedirs("../data/processed", exist_ok=True)

output_path = "../data/processed/imdb_movies_ratings.csv"

df_analysis.to_csv(
    output_path,
    index=False
)

print("Dataset saved successfully!")
print(output_path)
```
